
Intelligent Report Generation with Granular LLM-Based Editing
Azure OpenAI Implementation - Notebook Style

This notebook demonstrates a complete implementation of surgical document editing
using Azure OpenAI for targeted modifications with minimal token usage.

```python
# ============================================================================
# CELL 1: Setup and Installation
# ============================================================================

# Run this first to install required packages
"""
!pip install openai python-dotenv tiktoken
"""

import os
import re
import json
import hashlib
from typing import List, Dict, Optional, Tuple
from dataclasses import dataclass, field
from enum import Enum
import tiktoken
from openai import AzureOpenAI
from datetime import datetime

print("✓ Imports successful")
```

```python
# ============================================================================
# CELL 2: Configuration and Azure OpenAI Setup
# ============================================================================

# Azure OpenAI Configuration
# Replace these with your actual Azure OpenAI credentials
AZURE_CONFIG = {
    "api_key": os.getenv("AZURE_OPENAI_API_KEY", "your-api-key-here"),
    "api_version": os.getenv("AZURE_OPENAI_API_VERSION"),
    "azure_endpoint": os.getenv("AZURE_OPENAI_ENDPOINT"),
    "deployment_name": os.getenv("AZURE_OPENAI_CHAT_DEPLOYMENT_NAME"),  # Your deployment name
}

# Initialize Azure OpenAI client
client = AzureOpenAI(
    api_key=AZURE_CONFIG["api_key"],
    api_version=AZURE_CONFIG["api_version"],
    azure_endpoint=AZURE_CONFIG["azure_endpoint"]
)

# Initialize tokenizer for token counting
tokenizer = tiktoken.encoding_for_model("gpt-40-mini")

print("✓ Azure OpenAI client initialized")
print(f"  Endpoint: {AZURE_CONFIG['azure_endpoint']}")
print(f"  Deployment: {AZURE_CONFIG['deployment_name']}")

```

```python
# ============================================================================
# CELL 3: Data Structures
# ============================================================================

class ActionType(Enum):
    """Types of document editing actions"""
    ADD = "add"
    REMOVE = "remove"
    UPDATE = "update"
    EXPAND = "expand"
    REORDER = "reorder"
    UNKNOWN = "unknown"


@dataclass
class DocumentSection:
    """Represents a section in the document hierarchy"""
    id: str
    type: str  # h1, h2, h3, paragraph
    title: str
    content: str
    level: int
    start_line: int
    end_line: int
    children: List['DocumentSection'] = field(default_factory=list)
    parent: Optional['DocumentSection'] = None
    
    @property
    def full_content(self) -> str:
        """Get full content including title"""
        prefix = "#" * self.level
        return f"{prefix} {self.title}\n\n{self.content}"
    
    @property
    def word_count(self) -> int:
        return len(self.content.split())
    
    @property
    def token_count(self) -> int:
        return len(tokenizer.encode(self.content))
    
    def get_hash(self) -> str:
        """Generate hash for change detection"""
        return hashlib.md5(self.content.encode()).hexdigest()


@dataclass
class EditIntent:
    """Parsed user command intent"""
    action: ActionType
    target_section: Optional[str]
    position: Optional[Dict[str, str]]
    requirements: Optional[str]
    confidence: float
    
    def __str__(self):
        return f"Action: {self.action.value}, Target: {self.target_section}, Confidence: {self.confidence:.2%}"


@dataclass
class EditMetrics:
    """Performance metrics for an edit operation"""
    tokens_used: int
    tokens_full_regen: int
    token_reduction_pct: float
    latency_ms: int
    sections_modified: int
    total_sections: int
    action_type: str
    
    def __str__(self):
        return f"""
Edit Metrics:
  Token Reduction: {self.token_reduction_pct:.1f}%
  Tokens Used: {self.tokens_used} (vs {self.tokens_full_regen} full regen)
  Latency: {self.latency_ms}ms
  Sections Modified: {self.sections_modified}/{self.total_sections}
  Action: {self.action_type}
"""
```

```python
# ============================================================================
# CELL 4: Document Parser
# ============================================================================

class DocumentParser:
    """Parses documents into hierarchical section structure"""
    
    def __init__(self):
        self.sections = []
        
    def parse(self, document: str) -> List[DocumentSection]:
        """Parse document into structured sections"""
        lines = document.split('\n')
        sections = []
        section_stack = []
        current_content = []
        current_line = 0
        
        for i, line in enumerate(lines):
            heading_match = re.match(r'^(#{1,6})\s+(.+)$', line)
            
            if heading_match:
                # Save previous section content
                if section_stack:
                    section_stack[-1].content = '\n'.join(current_content).strip()
                    section_stack[-1].end_line = i - 1
                    current_content = []
                
                # Create new section
                level = len(heading_match.group(1))
                title = heading_match.group(2).strip()
                
                section = DocumentSection(
                    id=f"sec_{len(sections)}",
                    type=f"h{level}",
                    title=title,
                    content="",
                    level=level,
                    start_line=i,
                    end_line=i
                )
                
                # Handle hierarchy
                while section_stack and section_stack[-1].level >= level:
                    completed = section_stack.pop()
                    if section_stack:
                        section_stack[-1].children.append(completed)
                    else:
                        sections.append(completed)
                
                if section_stack:
                    section.parent = section_stack[-1]
                
                section_stack.append(section)
                
            else:
                # Accumulate content
                if line.strip():
                    current_content.append(line)
        
        # Handle remaining sections
        if section_stack:
            section_stack[-1].content = '\n'.join(current_content).strip()
            section_stack[-1].end_line = len(lines) - 1
            
            while section_stack:
                completed = section_stack.pop()
                if section_stack:
                    section_stack[-1].children.append(completed)
                else:
                    sections.append(completed)
        
        self.sections = sections
        return sections
    
    def find_section(self, sections: List[DocumentSection], title: str) -> Optional[DocumentSection]:
        """Find section by title with fuzzy matching"""
        title_lower = title.lower().strip()
        
        def search(secs):
            for section in secs:
                if title_lower in section.title.lower() or section.title.lower() in title_lower:
                    return section
                if section.children:
                    result = search(section.children)
                    if result:
                        return result
            return None
        
        return search(sections)
    
    def get_section_context(self, section: DocumentSection, max_tokens: int = 200) -> Dict[str, str]:
        """Get surrounding context for a section"""
        context = {}
        
        # Get previous sibling or parent content
        if section.parent and section.parent.children:
            idx = section.parent.children.index(section)
            if idx > 0:
                prev = section.parent.children[idx - 1]
                context['before'] = prev.content[:500]  # Limit context
        
        # Get next sibling content
        if section.parent and section.parent.children:
            idx = section.parent.children.index(section)
            if idx < len(section.parent.children) - 1:
                next_sec = section.parent.children[idx + 1]
                context['after'] = next_sec.content[:500]
        
        return context


print("✓ DocumentParser class defined")

```

```python
# ============================================================================
# CELL 5: Intent Recognition Engine
# ============================================================================

class IntentRecognizer:
    """Recognizes user intent from commands using Azure OpenAI"""
    
    def __init__(self, client: AzureOpenAI, deployment: str):
        self.client = client
        self.deployment = deployment
        
    def parse_command_rule_based(self, command: str) -> Optional[EditIntent]:
        """Fast rule-based parsing for common patterns"""
        cmd = command.lower().strip()
        
        # Remove patterns
        if re.search(r'\b(remove|delete)\b', cmd):
            match = re.search(r'(?:remove|delete)\s+(?:section\s+)?["\']?([^"\']+)["\']?', cmd, re.I)
            if match:
                return EditIntent(
                    action=ActionType.REMOVE,
                    target_section=match.group(1).strip(),
                    position=None,
                    requirements=None,
                    confidence=0.9
                )
        
        # Add patterns
        if re.search(r'\b(add|insert|create)\b', cmd):
            content_match = re.search(r'(?:add|insert|create)\s+(?:a\s+)?(?:section\s+)?(?:about|on|for)?\s*["\']?([^"\']+)["\']?', cmd, re.I)
            after_match = re.search(r'(?:after|below)\s+["\']?([^"\']+)["\']?', cmd, re.I)
            before_match = re.search(r'(?:before|above)\s+["\']?([^"\']+)["\']?', cmd, re.I)
            
            if content_match:
                position = {}
                if after_match:
                    position = {"after": after_match.group(1).strip()}
                elif before_match:
                    position = {"before": before_match.group(1).strip()}
                else:
                    position = {"end": True}
                
                return EditIntent(
                    action=ActionType.ADD,
                    target_section=None,
                    position=position,
                    requirements=content_match.group(1).strip(),
                    confidence=0.85
                )
        
        # Update patterns
        if re.search(r'\b(update|modify|change|rewrite)\b', cmd):
            target_match = re.search(r'(?:update|modify|change|rewrite)\s+(?:section\s+)?["\']?([^"\']+)["\']?', cmd, re.I)
            to_match = re.search(r'to\s+["\']?([^"\']+)["\']?', cmd, re.I)
            
            if target_match:
                return EditIntent(
                    action=ActionType.UPDATE,
                    target_section=target_match.group(1).strip(),
                    position=None,
                    requirements=to_match.group(1).strip() if to_match else None,
                    confidence=0.88
                )
        
        # Expand patterns
        if re.search(r'\b(expand|elaborate|add more)\b', cmd):
            match = re.search(r'(?:expand|elaborate|add more to)\s+(?:section\s+)?["\']?([^"\']+)["\']?', cmd, re.I)
            if match:
                return EditIntent(
                    action=ActionType.EXPAND,
                    target_section=match.group(1).strip(),
                    position=None,
                    requirements=None,
                    confidence=0.87
                )
        
        return None
    
    def parse_command_llm(self, command: str) -> EditIntent:
        """Use Azure OpenAI for complex command parsing"""
        system_prompt = """You are a command parser for a document editing system.
Parse the user's command and extract:
1. action: one of [add, remove, update, expand, reorder]
2. target_section: the section name to modify (if applicable)
3. position: where to apply the change (before/after a section, or "end")
4. requirements: what content to generate or how to modify
5. confidence: your confidence score (0.0 to 1.0)

Return ONLY a JSON object with these fields. No explanation."""

        try:
            response = self.client.chat.completions.create(
                model=self.deployment,
                messages=[
                    {"role": "system", "content": system_prompt},
                    {"role": "user", "content": f"Parse this command: {command}"}
                ],
                temperature=0.1,
                max_tokens=200
            )
            
            result = json.loads(response.choices[0].message.content)
            
            return EditIntent(
                action=ActionType(result.get("action", "unknown")),
                target_section=result.get("target_section"),
                position=result.get("position"),
                requirements=result.get("requirements"),
                confidence=result.get("confidence", 0.5)
            )
            
        except Exception as e:
            print(f"LLM parsing failed: {e}")
            return EditIntent(
                action=ActionType.UNKNOWN,
                target_section=None,
                position=None,
                requirements=None,
                confidence=0.0
            )
    
    def parse(self, command: str) -> EditIntent:
        """Parse command using rule-based first, fallback to LLM"""
        # Try rule-based first (fast and free)
        intent = self.parse_command_rule_based(command)
        
        # Fallback to LLM if confidence is low or unknown
        if intent is None or intent.confidence < 0.8:
            print("  Using LLM for intent recognition...")
            intent = self.parse_command_llm(command)
        else:
            print("  Using rule-based intent recognition...")
        
        return intent


print("✓ IntentRecognizer class defined")
```

```python
# ============================================================================
# CELL 6: Targeted Edit Orchestrator
# ============================================================================

class EditOrchestrator:
    """Orchestrates targeted edits using Azure OpenAI"""
    
    def __init__(self, client: AzureOpenAI, deployment: str):
        self.client = client
        self.deployment = deployment
        
    def generate_content(self, action: ActionType, context: Dict) -> str:
        """Generate content for a specific action with minimal context"""
        
        if action == ActionType.ADD:
            prompt = f"""Write a new section titled "{context['title']}" for a business report.

Context - this section will be inserted near:
{context.get('before', 'N/A')[:300]}

Requirements: {context.get('requirements', 'Comprehensive analysis')}

Write the section content (2-3 paragraphs, professional business tone)."""

        elif action == ActionType.EXPAND:
            prompt = f"""Expand this section with additional details and insights.

Original content:
{context['original'][:500]}

Add 2-3 more paragraphs that:
- Provide deeper analysis
- Include supporting details
- Maintain the same professional tone
- Build naturally on the existing content

Write only the NEW paragraphs to add (don't repeat the original)."""

        elif action == ActionType.UPDATE:
            prompt = f"""Rewrite this section with the following changes.

Original content:
{context['original'][:500]}

Changes required: {context.get('requirements', 'Update with current information')}

Write the complete updated section (2-3 paragraphs)."""

        else:
            return ""
        
        try:
            response = self.client.chat.completions.create(
                model=self.deployment,
                messages=[
                    {"role": "system", "content": "You are a professional business report writer."},
                    {"role": "user", "content": prompt}
                ],
                temperature=0.7,
                max_tokens=500
            )
            
            return response.choices[0].message.content.strip()
            
        except Exception as e:
            print(f"Content generation failed: {e}")
            return ""
    
    def smooth_transition(self, before: str, after: str) -> str:
        """Generate a smooth transition between sections"""
        prompt = f"""Create a brief transition sentence (1 sentence) between these two sections:

Section ending:
{before[-200:]}

Next section beginning:
{after[:200]}

Write only the transition sentence."""

        try:
            response = self.client.chat.completions.create(
                model=self.deployment,
                messages=[
                    {"role": "system", "content": "You are a professional editor ensuring document coherence."},
                    {"role": "user", "content": prompt}
                ],
                temperature=0.5,
                max_tokens=50
            )
            
            return response.choices[0].message.content.strip()
            
        except Exception as e:
            return ""


print("✓ EditOrchestrator class defined")
```

```python
# ============================================================================
# CELL 7: Main Document Editor
# ============================================================================

class GranularDocumentEditor:
    """Main editor that coordinates all components"""
    
    def __init__(self, azure_config: Dict):
        self.client = AzureOpenAI(
            api_key=azure_config["api_key"],
            api_version=azure_config["api_version"],
            azure_endpoint=azure_config["azure_endpoint"]
        )
        
        self.parser = DocumentParser()
        self.intent_recognizer = IntentRecognizer(
            self.client, 
            azure_config["deployment_name_mini"]  # Use mini for parsing
        )
        self.orchestrator = EditOrchestrator(
            self.client,
            azure_config["deployment_name"]  # Use full model for generation
        )
        
        self.document = ""
        self.sections = []
        self.history = []
        
    def load_document(self, document: str):
        """Load and parse a document"""
        self.document = document
        self.sections = self.parser.parse(document)
        self.history = [document]
        print(f"✓ Loaded document with {len(self.sections)} top-level sections")
        
    def count_tokens(self, text: str) -> int:
        """Count tokens in text"""
        return len(tokenizer.encode(text))
    
    def apply_edit(self, command: str) -> Tuple[str, EditMetrics]:
        """Apply an edit command to the document"""
        if not self.document:
            raise ValueError("No document loaded. Call load_document() first.")
        
        start_time = datetime.now()
        
        # Parse intent
        print(f"\n📝 Processing command: {command}")
        intent = self.intent_recognizer.parse(command)
        print(f"  {intent}")
        
        if intent.action == ActionType.UNKNOWN:
            raise ValueError(f"Could not understand command. Confidence: {intent.confidence:.2%}")
        
        # Calculate full regeneration cost
        full_regen_tokens = self.count_tokens(self.document) * 2  # Input + output
        tokens_used = 0
        sections_modified = 0
        
        # Execute action
        new_document = self.document
        lines = self.document.split('\n')
        
        if intent.action == ActionType.REMOVE:
            section = self.parser.find_section(self.sections, intent.target_section)
            if not section:
                raise ValueError(f"Section not found: {intent.target_section}")
            
            # Remove section
            before = '\n'.join(lines[:section.start_line])
            after = '\n'.join(lines[section.end_line + 1:])
            new_document = before + '\n\n' + after
            
            tokens_used = 50  # Minimal for removal
            sections_modified = 1
            
        elif intent.action == ActionType.ADD:
            # Generate new content
            context = {
                'title': intent.requirements,
                'requirements': intent.requirements,
            }
            
            if intent.position.get('after'):
                section = self.parser.find_section(self.sections, intent.position['after'])
                if section:
                    context['before'] = section.content
                    insert_line = section.end_line + 1
                else:
                    insert_line = len(lines)
            else:
                insert_line = len(lines)
            
            tokens_used += self.count_tokens(json.dumps(context))
            
            new_content = self.orchestrator.generate_content(ActionType.ADD, context)
            tokens_used += self.count_tokens(new_content)
            
            # Insert content
            heading = f"## {intent.requirements}\n\n"
            before = '\n'.join(lines[:insert_line])
            after = '\n'.join(lines[insert_line:])
            new_document = before + '\n\n' + heading + new_content + '\n\n' + after
            
            sections_modified = 1
            
        elif intent.action == ActionType.EXPAND:
            section = self.parser.find_section(self.sections, intent.target_section)
            if not section:
                raise ValueError(f"Section not found: {intent.target_section}")
            
            context = {
                'original': section.content
            }
            tokens_used += self.count_tokens(section.content)
            
            additional_content = self.orchestrator.generate_content(ActionType.EXPAND, context)
            tokens_used += self.count_tokens(additional_content)
            
            # Insert expanded content
            before = '\n'.join(lines[:section.end_line + 1])
            after = '\n'.join(lines[section.end_line + 1:])
            new_document = before + '\n\n' + additional_content + '\n\n' + after
            
            sections_modified = 1
            
        elif intent.action == ActionType.UPDATE:
            section = self.parser.find_section(self.sections, intent.target_section)
            if not section:
                raise ValueError(f"Section not found: {intent.target_section}")
            
            context = {
                'original': section.content,
                'requirements': intent.requirements
            }
            tokens_used += self.count_tokens(section.content)
            
            updated_content = self.orchestrator.generate_content(ActionType.UPDATE, context)
            tokens_used += self.count_tokens(updated_content)
            
            # Replace content
            before = '\n'.join(lines[:section.start_line + 1])  # Keep heading
            after = '\n'.join(lines[section.end_line + 1:])
            new_document = before + '\n\n' + updated_content + '\n\n' + after
            
            sections_modified = 1
        
        # Calculate metrics
        latency = (datetime.now() - start_time).total_seconds() * 1000
        token_reduction = ((full_regen_tokens - tokens_used) / full_regen_tokens * 100)
        
        metrics = EditMetrics(
            tokens_used=tokens_used,
            tokens_full_regen=full_regen_tokens,
            token_reduction_pct=token_reduction,
            latency_ms=int(latency),
            sections_modified=sections_modified,
            total_sections=len(self.sections),
            action_type=intent.action.value
        )
        
        # Update document
        self.document = new_document
        self.sections = self.parser.parse(new_document)
        self.history.append(new_document)
        
        return new_document, metrics
    
    def show_diff(self, old_doc: str, new_doc: str):
        """Show a simple diff between documents"""
        old_lines = old_doc.split('\n')
        new_lines = new_doc.split('\n')
        
        print("\n" + "="*60)
        print("DOCUMENT DIFF")
        print("="*60)
        
        # Simple line-by-line comparison
        max_len = max(len(old_lines), len(new_lines))
        changes = 0
        
        for i in range(max_len):
            old_line = old_lines[i] if i < len(old_lines) else ""
            new_line = new_lines[i] if i < len(new_lines) else ""
            
            if old_line != new_line:
                changes += 1
                if not old_line:
                    print(f"+ {new_line}")
                elif not new_line:
                    print(f"- {old_line}")
                else:
                    print(f"- {old_line}")
                    print(f"+ {new_line}")
        
        print(f"\n{changes} lines changed")
        print("="*60 + "\n")
    
    def undo(self) -> str:
        """Undo last edit"""
        if len(self.history) > 1:
            self.history.pop()
            self.document = self.history[-1]
            self.sections = self.parser.parse(self.document)
            print("✓ Undo successful")
            return self.document
        else:
            print("! Nothing to undo")
            return self.document


print("✓ GranularDocumentEditor class defined")
```

```python
# ============================================================================
# CELL 8: Example Usage - Initialize Editor
# ============================================================================

# Create editor instance
editor = GranularDocumentEditor(AZURE_CONFIG)

# Sample document for testing
SAMPLE_DOCUMENT = """# Executive Summary

This report analyzes the Q4 2024 performance of our product division. Overall revenue increased by 23% compared to Q3, driven primarily by strong adoption in the enterprise segment.

# Market Analysis

## Current Market Position

Our market share has grown from 12% to 15% in the last quarter. This growth is attributed to successful product launches and enhanced customer service initiatives.

## Competitive Landscape

Three main competitors continue to dominate the space. Company A holds 30%, Company B holds 25%, and Company C holds 18% of the market.

# Financial Performance

Revenue reached $12.3M in Q4, representing a 23% increase quarter-over-quarter. Operating margins improved to 18%, up from 15% in Q3.

## Cost Structure

Operating expenses were well-controlled at $10.1M, resulting in strong profitability. Marketing spend increased by 15% to support product launches.

# Recommendations

We recommend increasing investment in the enterprise segment by 30% and expanding our sales team by 10 additional representatives in Q1 2025."""

# Load the document
editor.load_document(SAMPLE_DOCUMENT)

print("\n📄 Document Structure:")
print("-" * 60)
for i, section in enumerate(editor.sections):
    print(f"{i+1}. {section.title} (Level {section.level})")
    print(f"   Words: {section.word_count}, Tokens: {section.token_count}")
    for j, child in enumerate(section.children):
        print(f"   {i+1}.{j+1}. {child.title} (Level {child.level})")
```

```python

# ============================================================================
# CELL 14: Interactive Mode - Try Your Own Commands
# ============================================================================

def interactive_mode(editor: GranularDocumentEditor):
    """Run editor in interactive mode"""
    print("\n" + "="*80)
    print("INTERACTIVE MODE")
    print("="*80)
    print("\nAvailable commands:")
    print("  - remove/delete section <name>")
    print("  - add section about <topic> after <section>")
    print("  - expand <section> with more details")
    print("  - update <section> to <new requirements>")
    print("  - show document")
    print("  - undo")
    print("  - quit")
    print("\n" + "="*80 + "\n")
    
    while True:
        try:
            command = input("\n💬 Enter command: ").strip()
            
            if command.lower() in ['quit', 'exit', 'q']:
                print("👋 Goodbye!")
                break
            elif command.lower() == 'show document':
                print("\n" + "="*60)
                print(editor.document)
                print("="*60)
            elif command.lower() == 'undo':
                editor.undo()
            elif command:
                new_doc, metrics = editor.apply_edit(command)
                print(metrics)
                
        except Exception as e:
            print(f"❌ Error: {e}")
            print("Try a different command or type 'quit' to exit.")

# Uncomment to run interactive mode:
# interactive_mode(editor)
```

```python
# ============================================================================
# CELL 15: Save/Load Functions (COMPLETE)
# ============================================================================

def save_document(editor: GranularDocumentEditor, filename: str):
    """Save current document to file"""
    with open(filename, 'w', encoding='utf-8') as f:
        f.write(editor.document)
    print(f"✓ Document saved to {filename}")

def load_from_file(editor: GranularDocumentEditor, filename: str):
    """Load document from file"""
    with open(filename, 'r', encoding='utf-8') as f:
        document = f.read()
    editor.load_document(document)
    print(f"✓ Document loaded from {filename}")

def save_history(editor: GranularDocumentEditor, directory: str):
    """Save all document versions"""
    import os
    os.makedirs(directory, exist_ok=True)
    
    for i, version in enumerate(editor.history):
        filename = os.path.join(directory, f"version_{i}.md")
        with open(filename, 'w', encoding='utf-8') as f:
            f.write(version)
    
    print(f"✓ Saved {len(editor.history)} versions to {directory}")

# Example usage:
# save_document(editor, "my_report.md")
# load_from_file(editor, "my_report.md")
# save_history(editor, "./history")

print("✓ Save/Load functions defined")
```

```python

# ============================================================================
# CELL 16: Advanced Features - Batch Operations
# ============================================================================

class BatchEditor:
    """Handle multiple edit operations efficiently"""
    
    def __init__(self, editor: GranularDocumentEditor):
        self.editor = editor
        self.batch_queue = []
        
    def queue_edit(self, command: str):
        """Add command to batch queue"""
        self.batch_queue.append(command)
        print(f"✓ Queued: {command}")
        
    def execute_batch(self) -> List[EditMetrics]:
        """Execute all queued edits"""
        print(f"\n{'='*80}")
        print(f"EXECUTING BATCH: {len(self.batch_queue)} commands")
        print(f"{'='*80}\n")
        
        all_metrics = []
        
        for i, command in enumerate(self.batch_queue, 1):
            print(f"\n[{i}/{len(self.batch_queue)}] {command}")
            try:
                _, metrics = self.editor.apply_edit(command)
                all_metrics.append(metrics)
                print(f"✓ Success - {metrics.token_reduction_pct:.1f}% token reduction")
            except Exception as e:
                print(f"✗ Failed: {e}")
        
        self.batch_queue.clear()
        
        # Show aggregate metrics
        if all_metrics:
            total_tokens = sum(m.tokens_used for m in all_metrics)
            total_full_regen = sum(m.tokens_full_regen for m in all_metrics)
            avg_reduction = (total_full_regen - total_tokens) / total_full_regen * 100
            
            print(f"\n{'='*80}")
            print("BATCH SUMMARY")
            print(f"{'='*80}")
            print(f"Total Operations: {len(all_metrics)}")
            print(f"Total Tokens Used: {total_tokens}")
            print(f"vs Full Regeneration: {total_full_regen}")
            print(f"Average Token Reduction: {avg_reduction:.1f}%")
            print(f"{'='*80}\n")
        
        return all_metrics
    
    def clear_queue(self):
        """Clear the batch queue"""
        self.batch_queue.clear()
        print("✓ Batch queue cleared")

# Example usage:
# batch = BatchEditor(editor)
# batch.queue_edit("remove section X")
# batch.queue_edit("add section about Y after Z")
# batch.queue_edit("expand section W")
# batch.execute_batch()

print("✓ BatchEditor class defined")
```

```python

# ============================================================================
# CELL 17: Quality Assurance - Coherence Checker
# ============================================================================

class CoherenceChecker:
    """Check document coherence after edits"""
    
    def __init__(self, client: AzureOpenAI, deployment: str):
        self.client = client
        self.deployment = deployment
        
    def check_transition(self, section1: DocumentSection, section2: DocumentSection) -> Dict:
        """Check if transition between sections is smooth"""
        prompt = f"""Analyze the transition between these two sections.

Section 1 ending:
{section1.content[-300:]}

Section 2 beginning:
{section2.content[:300]}

Rate the transition quality (0-10) and explain any issues.
Return JSON: {{"score": 0-10, "issues": ["list of issues"], "suggestion": "how to improve"}}"""

        try:
            response = self.client.chat.completions.create(
                model=self.deployment,
                messages=[
                    {"role": "system", "content": "You are a document coherence analyst."},
                    {"role": "user", "content": prompt}
                ],
                temperature=0.3,
                max_tokens=200
            )
            
            return json.loads(response.choices[0].message.content)
        except Exception as e:
            return {"score": 5, "issues": [str(e)], "suggestion": "Manual review needed"}
    
    def check_document_coherence(self, editor: GranularDocumentEditor) -> Dict:
        """Check overall document coherence"""
        results = {
            "overall_score": 0,
            "transitions": [],
            "issues": []
        }
        
        sections = editor.sections
        scores = []
        
        print("\n" + "="*60)
        print("COHERENCE ANALYSIS")
        print("="*60)
        
        # Check transitions between consecutive sections
        for i in range(len(sections) - 1):
            result = self.check_transition(sections[i], sections[i+1])
            scores.append(result["score"])
            results["transitions"].append({
                "from": sections[i].title,
                "to": sections[i+1].title,
                "score": result["score"]
            })
            
            if result["score"] < 7:
                results["issues"].append({
                    "location": f"{sections[i].title} → {sections[i+1].title}",
                    "problems": result["issues"],
                    "suggestion": result["suggestion"]
                })
                print(f"⚠ {sections[i].title} → {sections[i+1].title}: {result['score']}/10")
            else:
                print(f"✓ {sections[i].title} → {sections[i+1].title}: {result['score']}/10")
        
        results["overall_score"] = sum(scores) / len(scores) if scores else 0
        
        print(f"\nOverall Coherence Score: {results['overall_score']:.1f}/10")
        print("="*60 + "\n")
        
        return results

# Example usage:
# checker = CoherenceChecker(client, AZURE_CONFIG["deployment_name_mini"])
# coherence_report = checker.check_document_coherence(editor)

print("✓ CoherenceChecker class defined")
```

```python

# ============================================================================
# CELL 18: Analytics - Token Usage Dashboard
# ============================================================================

class TokenAnalytics:
    """Track and visualize token usage"""
    
    def __init__(self):
        self.operations = []
        
    def log_operation(self, metrics: EditMetrics):
        """Log an operation's metrics"""
        self.operations.append({
            "timestamp": datetime.now(),
            "action": metrics.action_type,
            "tokens_used": metrics.tokens_used,
            "tokens_saved": metrics.tokens_full_regen - metrics.tokens_used,
            "reduction_pct": metrics.token_reduction_pct,
            "latency_ms": metrics.latency_ms
        })
    
    def get_summary(self) -> Dict:
        """Get summary statistics"""
        if not self.operations:
            return {"message": "No operations logged yet"}
        
        total_tokens = sum(op["tokens_used"] for op in self.operations)
        total_saved = sum(op["tokens_saved"] for op in self.operations)
        avg_reduction = sum(op["reduction_pct"] for op in self.operations) / len(self.operations)
        avg_latency = sum(op["latency_ms"] for op in self.operations) / len(self.operations)
        
        # Cost calculation (GPT-4 pricing)
        cost_actual = (total_tokens / 1_000_000) * 20  # Average of input/output
        cost_full_regen = ((total_tokens + total_saved) / 1_000_000) * 20
        cost_saved = cost_full_regen - cost_actual
        
        return {
            "total_operations": len(self.operations),
            "total_tokens_used": total_tokens,
            "total_tokens_saved": total_saved,
            "avg_reduction_pct": avg_reduction,
            "avg_latency_ms": avg_latency,
            "cost_actual_usd": cost_actual,
            "cost_saved_usd": cost_saved,
            "actions": {action: sum(1 for op in self.operations if op["action"] == action)
                       for action in set(op["action"] for op in self.operations)}
        }
    
    def print_dashboard(self):
        """Print analytics dashboard"""
        summary = self.get_summary()
        
        if "message" in summary:
            print(summary["message"])
            return
        
        print("\n" + "="*80)
        print("TOKEN USAGE DASHBOARD")
        print("="*80)
        print(f"\n📊 Operations: {summary['total_operations']}")
        print(f"\n💰 Cost Analysis:")
        print(f"   Actual Cost:    ${summary['cost_actual_usd']:.4f}")
        print(f"   Cost Saved:     ${summary['cost_saved_usd']:.4f}")
        print(f"   Savings Rate:   {summary['avg_reduction_pct']:.1f}%")
        print(f"\n🎯 Token Usage:")
        print(f"   Tokens Used:    {summary['total_tokens_used']:,}")
        print(f"   Tokens Saved:   {summary['total_tokens_saved']:,}")
        print(f"\n⚡ Performance:")
        print(f"   Avg Latency:    {summary['avg_latency_ms']:.0f}ms")
        print(f"\n📈 Actions Breakdown:")
        for action, count in summary['actions'].items():
            print(f"   {action.capitalize():12} {count}")
        print("="*80 + "\n")

# Example usage:
# analytics = TokenAnalytics()
# _, metrics = editor.apply_edit("remove section X")
# analytics.log_operation(metrics)
# analytics.print_dashboard()

print("✓ TokenAnalytics class defined")
```

```python

# ============================================================================
# CELL 19: Enhanced Interactive Mode with Analytics
# ============================================================================

def interactive_mode_advanced(editor: GranularDocumentEditor):
    """Enhanced interactive mode with analytics and quality checks"""
    
    analytics = TokenAnalytics()
    batch = BatchEditor(editor)
    checker = CoherenceChecker(client, AZURE_CONFIG["deployment_name_mini"])
    
    print("\n" + "="*80)
    print("ADVANCED INTERACTIVE MODE")
    print("="*80)
    print("\n📝 Available commands:")
    print("   edit commands:")
    print("     - remove/delete section <name>")
    print("     - add section about <topic> after <section>")
    print("     - expand <section> with more details")
    print("     - update <section> to <new requirements>")
    print("\n   batch commands:")
    print("     - batch add <command>         (add to batch queue)")
    print("     - batch run                   (execute batch)")
    print("     - batch clear                 (clear queue)")
    print("\n   utility commands:")
    print("     - show document               (display current document)")
    print("     - show sections               (show document structure)")
    print("     - check coherence             (analyze document quality)")
    print("     - analytics                   (show token usage stats)")
    print("     - undo                        (revert last change)")
    print("     - save <filename>             (save document)")
    print("     - quit                        (exit)")
    print("\n" + "="*80 + "\n")
    
    while True:
        try:
            command = input("\n💬 Enter command: ").strip()
            
            if not command:
                continue
            
            # Exit commands
            if command.lower() in ['quit', 'exit', 'q']:
                print("👋 Goodbye!")
                break
            
            # Document viewing
            elif command.lower() == 'show document':
                print("\n" + "="*60)
                print(editor.document)
                print("="*60)
            
            elif command.lower() == 'show sections':
                print("\n" + "="*60)
                print("DOCUMENT STRUCTURE")
                print("="*60)
                for i, section in enumerate(editor.sections, 1):
                    print(f"\n{i}. {section.title} (Level {section.level})")
                    print(f"   Words: {section.word_count}, Tokens: {section.token_count}")
                    for j, child in enumerate(section.children, 1):
                        print(f"   {i}.{j}. {child.title}")
                print("="*60)
            
            # Quality checks
            elif command.lower() == 'check coherence':
                print("\n🔍 Analyzing document coherence...")
                report = checker.check_document_coherence(editor)
                if report["issues"]:
                    print("\n⚠ Issues found:")
                    for issue in report["issues"]:
                        print(f"\nLocation: {issue['location']}")
                        print(f"Problems: {', '.join(issue['problems'])}")
                        print(f"Suggestion: {issue['suggestion']}")
            
            # Analytics
            elif command.lower() == 'analytics':
                analytics.print_dashboard()
            
            # Batch operations
            elif command.lower().startswith('batch '):
                parts = command.split(maxsplit=2)
                if len(parts) < 2:
                    print("❌ Invalid batch command")
                    continue
                
                batch_cmd = parts[1].lower()
                
                if batch_cmd == 'add' and len(parts) == 3:
                    batch.queue_edit(parts[2])
                elif batch_cmd == 'run':
                    metrics_list = batch.execute_batch()
                    for m in metrics_list:
                        analytics.log_operation(m)
                elif batch_cmd == 'clear':
                    batch.clear_queue()
                else:
                    print("❌ Invalid batch subcommand (use: add, run, clear)")
            
            # Save
            elif command.lower().startswith('save '):
                filename = command.split(maxsplit=1)[1]
                save_document(editor, filename)
            
            # Undo
            elif command.lower() == 'undo':
                editor.undo()
            
            # Edit operations
            else:
                new_doc, metrics = editor.apply_edit(command)
                analytics.log_operation(metrics)
                print(metrics)
                
        except KeyboardInterrupt:
            print("\n\n👋 Interrupted. Goodbye!")
            break
        except Exception as e:
            print(f"❌ Error: {e}")
            print("Try a different command or type 'quit' to exit.")

# Uncomment to run:
# interactive_mode_advanced(editor)

print("✓ Advanced interactive mode defined")

```
